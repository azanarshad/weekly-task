```python
import numpy as np 
              
marks = np.array([
    #m  , p, c
    [70, 80, 90],   # s1
    [60, 75, 85],   # s2
    [65, 70, 80],   # s3
    [75, 85, 95],   # s4
    [80, 90, 100]   # s5
])

# average marks of each student

average = np.mean(marks, axis=1 )

print("Avergae marks of each student are:", average)

# highest marks in each subject

h_maths = np.max(marks,axis=0)
print("Highest marks in maths:", h_maths)

#over class average
class_average = np.mean(marks)
print("Class average is:", class_average)
```

```python
import pandas as pd

dataframe = pd.DataFrame({"Name":["Azan", "Bilal", "Sana", "Ali", "Zara"],"Age":[18,19,21,23,20],"marks":[85,90,34,23,50]})
print(dataframe)
```

```python
# discply students having above average marks
above_average = dataframe[dataframe["marks"] > dataframe["marks"].mean()]
print("Students having above average marks:\n", above_average)
```

```python
#add a new column result 
dataframe["Result"] = ["Pass" if marks >= 50 else "Fail" for marks in dataframe["marks"]]
print("Dataframe with Result column:\n", dataframe)
```
