```python
import numpy as np 


x = np.array(([1, 2, 3],[4,5,6] ))

np.ndim(x) # number of dimensions
```

```python
identity_matrix = np.eye(2) # identity matrix of size 2x2
print(identity_matrix)
```

```python
full_matrix = np.full((2,3),7) # 2x2 matrix filled with 7s
print(full_matrix)
```

```python
floating_array = np.array(([1, 2],[3, 4])) # create a 2D array

float_matrix = np.float16(floating_array)# convert to float
print(float_matrix)
```

```python
arranged_array = np.arange(100,600,100)  # array from 100 to 500 with step of 100
print(arranged_array)
```

```python
y = np.array(([10,20],[30,40]))
z = np.array(([5,15],[25,35]))

result = y * z 
print(result)
```
